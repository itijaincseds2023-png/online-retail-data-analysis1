```python
from openpyxl import load_workbook
from collections import Counter

file_path = "/content/retail_data/online_retail_II.xlsx"

wb = load_workbook(file_path, read_only=True, data_only=True)

for ws in wb.worksheets:
    missing = Counter()

    for row in ws.iter_rows(min_row=2, values_only=True):
        for i, value in enumerate(row):
            if value is None or value == "":
                missing[i] += 1

    print("\n", ws.title)
    print("-" * 30)

    for i, count in missing.items():
        print(f"{ws.cell(1, i+1).value}: {count} missing values")

wb.close()
```

```python
import hashlib
from openpyxl import load_workbook

file_path = "/content/retail_data/online_retail_II.xlsx"

wb = load_workbook(file_path, read_only=True, data_only=True)

for ws in wb.worksheets:
    seen = set()
    duplicates = 0

    for row in ws.iter_rows(min_row=2, values_only=True):
        row_text = "|".join("" if x is None else str(x) for x in row)
        row_hash = hashlib.md5(row_text.encode()).hexdigest()

        if row_hash in seen:
            duplicates += 1
        else:
            seen.add(row_hash)

    print(ws.title, "→", duplicates, "exact duplicate rows")

wb.close()
```

```python
from openpyxl import load_workbook

file_path = "/content/retail_data/online_retail_II.xlsx"

wb = load_workbook(file_path, read_only=True, data_only=True)

for ws in wb.worksheets:
    total = 0
    negative = 0
    zero = 0

    # Quantity column = 4th column
    for row in ws.iter_rows(min_row=2, values_only=True):
        quantity = row[3]

        if quantity is not None:
            total += 1

            if quantity < 0:
                negative += 1
            elif quantity == 0:
                zero += 1

    print(f"\n{ws.title}")
    print(f"Total quantity records: {total}")
    print(f"Negative quantity: {negative}")
    print(f"Zero quantity: {zero}")

wb.close()
```

```python
from openpyxl import load_workbook

file_path = "/content/retail_data/online_retail_II.xlsx"

wb = load_workbook(file_path, read_only=True, data_only=True)

for ws in wb.worksheets:
    negative = 0
    cancelled_invoice = 0

    for row in ws.iter_rows(min_row=2, values_only=True):
        invoice = row[0]
        quantity = row[3]

        if quantity is not None and quantity < 0:
            negative += 1

            if invoice is not None and str(invoice).startswith("C"):
                cancelled_invoice += 1

    print(f"\n{ws.title}")
    print(f"Negative quantity: {negative}")
    print(f"Negative quantity with C invoice: {cancelled_invoice}")

wb.close()
```

```python
from openpyxl import load_workbook

file_path = "/content/retail_data/online_retail_II.xlsx"

wb = load_workbook(file_path, read_only=True, data_only=True)

for ws in wb.worksheets:
    total_invoices = set()
    cancelled_invoices = set()

    for row in ws.iter_rows(min_row=2, values_only=True):
        invoice = row[0]

        if invoice is not None:
            invoice = str(invoice)
            total_invoices.add(invoice)

            if invoice.startswith("C"):
                cancelled_invoices.add(invoice)

    print(f"\n{ws.title}")
    print(f"Unique invoices: {len(total_invoices)}")
    print(f"Cancellation invoices: {len(cancelled_invoices)}")

wb.close()
```

```python
import pandas as pd

file_path = "/content/retail_data/online_retail_II.xlsx"

# Dono sheets read karna
df_2009 = pd.read_excel(file_path, sheet_name="Year 2009-2010")
df_2010 = pd.read_excel(file_path, sheet_name="Year 2010-2011")

# Dono years ko combine karna
df = pd.concat([df_2009, df_2010], ignore_index=True)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print(df.columns.tolist())
```

```python
# Duplicate rows check
duplicates = df.duplicated().sum()

print("Duplicate rows:", duplicates)

# Duplicate rows remove
df = df.drop_duplicates().reset_index(drop=True)

print("Rows after removing duplicates:", len(df))
```

```python
# Check missing values in each column
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Handle missing descriptions
df["Description"] = df["Description"].fillna("Unknown")

# Check again
print("Missing Description:", df["Description"].isnull().sum())
print("Missing Customer ID:", df["Customer ID"].isnull().sum())
```

```python
# Check data types
print(df.dtypes)
```

```python
# Create Sales Value column
df["Sales Value"] = df["Quantity"] * df["Price"]

# Check the result
print(df[["Quantity", "Price", "Sales Value"]].head())
```

```python
# Identify cancellation/return transactions
df["Is_Cancellation"] = df["Quantity"] < 0

# Check counts
print("Cancellation/Return rows:", df["Is_Cancellation"].sum())
print("Normal sales rows:", (~df["Is_Cancellation"]).sum())
```

```python
# Create Revenue column
df["Revenue"] = df["Quantity"] * df["Price"]

# Check the result
print(df[["Quantity", "Price", "Revenue", "Is_Cancellation"]].head(10))
```

```python
# Final data quality check

print("Total rows:", len(df))
print("Total columns:", len(df.columns))
print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

print("\nColumns:")
print(df.columns.tolist())
```

```python
# Save processed data as CSV
output_path = "/content/Processed_Online_Retail.csv"

df.to_csv(output_path, index=False)

print("Processed file saved successfully!")
print(output_path)
```

```python
# Monthly revenue analysis

df["Month"] = df["InvoiceDate"].dt.to_period("M").astype(str)

monthly_revenue = (
    df.groupby("Month")["Revenue"]
      .sum()
      .reset_index()
)

print(monthly_revenue.to_string(index=False))
```

```python
# Top 10 products by revenue

top_products = (
    df.groupby("Description")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

print(top_products.to_string(index=False))
```

```python
top_countries = (
    df.groupby("Country")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

print(top_countries.to_string(index=False))
```

```python
cancellation_by_month = (
    df[df["Is_Cancellation"] == True]
    .groupby("Month")
    .size()
    .sort_values(ascending=False)
    .head(10)
    .reset_index(name="Cancellation_Count")
)

print(cancellation_by_month.to_string(index=False))
```

```python
top_customers = (
    df[df["Customer ID"].notna()]
    .groupby("Customer ID")["Revenue"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

print(top_customers.to_string(index=False))
```

```python
nov_dec = (
    df[df["Month"].str.endswith(("-11", "-12"))]
    .groupby("Month")["Revenue"]
    .sum()
    .reset_index()
)

print(nov_dec.to_string(index=False))
```

```python
# Dashboard ke liye summary data

monthly = (
    df.groupby("Month")["Revenue"]
      .sum()
      .reset_index()
)

countries = (
    df.groupby("Country")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

products = (
    df.groupby("Description")["Revenue"]
      .sum()
      .sort_values(ascending=False)
      .head(20)
      .reset_index()
)

cancellations = (
    df[df["Is_Cancellation"] == True]
      .groupby("Month")
      .size()
      .reset_index(name="Cancellation_Count")
)

print("Monthly:", monthly.shape)
print("Countries:", countries.shape)
print("Products:", products.shape)
print("Cancellations:", cancellations.shape)
```

```python
import pandas as pd

with pd.ExcelWriter("/content/Dashboard_Data.xlsx") as writer:
    monthly.to_excel(writer, sheet_name="Monthly Revenue", index=False)
    countries.to_excel(writer, sheet_name="Country Revenue", index=False)
    products.to_excel(writer, sheet_name="Top Products", index=False)
    cancellations.to_excel(writer, sheet_name="Cancellations", index=False)

print("Dashboard_Data.xlsx created successfully!")
```
